**The FinanceToolkit** is an open-source toolkit in which all relevant financial methods (500+) are written down in the most simplistic way allowing for complete transparency of the calculation method. This allows you to not have to rely on metrics from other providers and, given a financial statement, allow for efficient manual calculations. This leads to one uniform method of calculation being applied that is available and understood by everyone.

````python
import numpy as np
import pandas as pd

from financetoolkit import Toolkit

API_KEY = "FINANCIAL_MODELING_PREP_API_KEY"
````

**Initializing only is required once.** This is the case for any function so once you have obtained a balance sheet statement, it will be stored accordingly which means that requests to FinancialModelingPrep, the source used in these examples, are kept to a minimum. Note that in this example annual data is used but by adding `quarterly=True` to the Toolkit initialization, quarterly data can also be collected. Note that this requires a Premium subscription from FMP.

````python
# Initialize the Toolkit with company tickers
companies = Toolkit(
    ["AAPL", "TSLA", "META", "AMZN"], api_key=API_KEY, start_date="2005-01-01"
)
````

After initialization of `Toolkit`, you can get access to the Models module which includes a variety of well-known models. This can be done by calling the `models` property. For example, it is possible to perform an (extended) Dupont analysis for each company.

````python
companies.models.get_extended_dupont_analysis()
````

Showing the growth is possible with the `growth=True` parameter which also includes a lag parameter in which 1 is the default. It is also possible to define multiple lags. As an example:

````python
companies.models.get_weighted_average_cost_of_capital(growth=True, lag=[1, 2, 3])
````

Other than that, there are many other models such as the Altman Z-Score and the Piotrivski F-Score.

````python
display(companies.models.get_altman_z_score().loc["AMZN"])

display(companies.models.get_piotroski_score().loc["AAPL"])

companies.models.get_piotroski_score().loc[:, "Piotroski Score", :].T.plot.bar(
    figsize=(15, 3), rot=0, title="Piotroski Score", colormap="tab20c", ylim=(0, 9)
)
````

Two related valuation views: the **Enterprise Value breakdown** decomposes EV into its components (market cap, debt, preferred equity, minority interest, cash), and the **Intrinsic Valuation** runs a full DCF given a growth rate, perpetual growth rate and WACC to estimate a fair share price.

````python
display(companies.models.get_enterprise_value_breakdown())

display(companies.models.get_intrinsic_valuation(0.05, 0.025, 0.094))
````

The base **DuPont Analysis** breaks Return on Equity down into three components — Net Profit Margin, Asset Turnover and Equity Multiplier — showing whether a company's ROE is driven by profitability, efficient asset use, or leverage. The **Present Value of Growth Opportunities (PVGO)** splits the current share price into the value already generated by existing operations versus the value the market is pricing in for future growth.

````python
companies.models.get_dupont_analysis(standardize=True).loc["AAPL"].T.plot(
    figsize=(15, 3),
    title="Dupont Analysis for Apple (standardized)",
    grid=True,
    linestyle="-",
    linewidth=2,
    xlabel="Date",
    ylabel="Return on Equity",
)

display(companies.models.get_present_value_of_growth_opportunities())
````

The **Weighted Average Cost of Capital (WACC)** — already introduced above via its growth over multiple lags — can also be plotted directly over time to see how a company's blended cost of equity and debt financing has moved.

````python
companies.models.get_weighted_average_cost_of_capital(standardize=True,show_full_results=True).loc['AMZN'].T.plot(
    figsize=(15, 3),
    title="Weighted Average Cost of Capital for Amazon (standardized)",
    grid=True,
    linestyle="-",
    linewidth=2,
    xlabel="Date",
    ylabel="WACC",
)
````

The **Beneish M-Score** is a probabilistic model for detecting earnings manipulation, a companion to the Altman Z-Score and Piotroski F-Score above. A score above roughly **-1.78** suggests the company is more likely to be manipulating earnings.

````python
companies.models.get_beneish_m_score().loc["META"]
````

The **Sustainable Growth Rate (SGR)** and **Internal Growth Rate (IGR)** estimate the fastest a company can grow revenue without external financing — SGR assumes leverage stays constant, IGR assumes no additional debt is raised at all, so IGR is always the more conservative (lower) of the two.

````python
display(companies.models.get_sustainable_growth_rate())

display(companies.models.get_internal_growth_rate())
````

Finally, the **Graham Number** is Benjamin Graham's conservative estimate of a stock's fair value, intended as an upper bound on what a defensive investor should pay. It's only meaningful for profitable companies with a positive book value — negative earnings or book value produce NaN.

````python
graham_number = companies.models.get_graham_number().iloc[:, -1]
current_price = companies.get_historical_data()["Adj Close"].iloc[-1].drop("Benchmark")

pd.DataFrame({"Graham Number": graham_number, "Current Price": current_price}).plot.bar(
    figsize=(15, 3),
    title="Graham Number vs Current Stock Price",
    grid=True,
    rot=0,
    ylabel="Price",
)
````

It is also possible to directly call the models that are part of the Models module. For example, The Dupont analysis functionality was filled with actual data from Apple and therefore it returns the results that closely match their actual metrics given the reporting dates in October.

````python
from financetoolkit.models import dupont_model

years = [2018, 2019, 2020, 2021, 2022]

dupont_model.get_dupont_analysis(
    net_income=pd.Series(
        [59531000000, 55256000000, 57411000000, 94680000000, 99803000000], index=years
    ),
    total_revenue=pd.Series(
        [265595000000, 260174000000, 274515000000, 365817000000, 394328000000],
        index=years,
    ),
    average_total_assets=pd.Series(
        [np.nan, 365725000000, 338516000000, 323888000000, 351002000000],
        index=years,
    ),
    average_total_equity=pd.Series(
        [np.nan, 107147000000, 90488000000, 65339000000, 63090000000], index=years
    ),
)
````
